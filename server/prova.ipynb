```python
groups = root.xpath("./svg:g", namespaces=NSMAP)
layers = []
for g in groups:
    for layer in g:
        layers.append(layer)
```

```python
def translate_element(elem, deltaX, deltaY):
    """
    Translate an SVG element by deltaX, deltaY by updating its transform.
    """
    # Get existing transform (if any)
    existing = elem.get("transform", "")
    
    # Prepend the new translation
    translate = f"translate({deltaX},{deltaY})"
    
    if existing:
        # Combine with existing transforms
        elem.set("transform", translate + " " + existing)
    else:
        elem.set("transform", translate)
    return elem
```
