```python
def get_layers(filename):
    from lxml import etree

    SVG_NS = "http://www.w3.org/2000/svg"
    NSMAP = {"svg": SVG_NS}

    # Load SVG
    tree = etree.parse(filename)
    root = tree.getroot()
    groups = root.xpath("./svg:g", namespaces=NSMAP)
    layers = []
    for g in groups:
        for layer in g:
            layers.append(layer)
    return layers

# layers = get_layers('./outputs/marti/output-marti.svg')
```
